# AgroSynapse EDA

Exploratory data analysis for cleaned survey data.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

DATA_PATH = 'data/processed/clean_data.csv'
df = pd.read_csv(DATA_PATH)
df.head()
```

```python
df.info()
```

```python
df.describe(include='all').transpose().head(20)
```

```python
missing = df.isna().mean().sort_values(ascending=False)
missing.head(20)
```

```python
numeric_cols = df.select_dtypes(include='number').columns
df[numeric_cols].hist(figsize=(12, 8))
plt.tight_layout()
```

```python
if 'yield_tons' in df.columns and 'farm_size_ha' in df.columns:
    sns.scatterplot(data=df, x='farm_size_ha', y='yield_tons')
    plt.title('Yield vs Farm Size')
```
